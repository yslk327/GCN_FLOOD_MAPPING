```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# === Charger le fichier CSV exporté depuis GEE ===
# Remplace le chemin ci-dessous par le chemin local vers ton fichier
csv_file = 'Samples_DL_Matam_10_2024_700_points.csv'
df = pd.read_csv(csv_file)

# === Vérification des colonnes ===
print("Colonnes disponibles :", df.columns.tolist())
print(df['classvalue'].value_counts())

# === Tracer les histogrammes de NDWI par classe ===
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='NDWI', hue='classvalue', bins=50, palette='Set1', element='step')
plt.title('Distribution des valeurs NDWI par classe')
plt.xlabel('NDWI')
plt.ylabel('Nombre de pixels')
plt.legend(title='Classe', labels=['Eau (1)','Non Eau (0)'])
plt.grid(True)
plt.tight_layout()
plt.show()
```
